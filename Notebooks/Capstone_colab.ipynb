```python
import yfinance as yf
import pandas as pd
import time
import os

start_date = "2020-01-01"
end_date = "2021-03-31"

tickers = {
    "BA": "Boeing",
    "SLX": "Steel ETF",
    "CL=F": "Crude Oil",
    "ITA": "Aerospace ETF",
    "XLI": "Industrial ETF",
    "^TNX": "10Y Yield",
    "^VIX": "Volatility Index",
    "DX-Y.NYB": "USD Index",
    "IYT": "Transport ETF",
    "JETS": "Airline ETF"
}

data_frames = []

for ticker, label in tickers.items():
    file_name = f"{label.replace(' ', '_')}.csv"

    if os.path.exists(file_name):
        print(f"Loading cached data for {label} from {file_name}")
        df = pd.read_csv(file_name, index_col=0, parse_dates=True)
    else:
        try:
            print(f"Downloading {label} ({ticker})...")
            df_all = yf.download(ticker, start=start_date, end=end_date)

            if "Adj Close" in df_all.columns:
                df = df_all[["Adj Close"]].rename(columns={"Adj Close": label})
                print(f"Using 'Adj Close' for {label}")
            elif "Close" in df_all.columns:
                df = df_all[["Close"]].rename(columns={"Close": label})
                print(f" 'Adj Close' not available — using 'Close' for {label}")
            else:
                print(f"No usable price column for {ticker}. Skipping.")
                continue

            df.to_csv(file_name)
            time.sleep(2)

        except Exception as e:
            print(f" Failed to download {ticker}: {e}")
            continue

    data_frames.append(df)

# Combine all successfully loaded series
if data_frames:
    combined_df = pd.concat(data_frames, axis=1)
    combined_df.to_csv("boeing_exo_variables_combined.csv")
    print("\n Combined DataFrame:")
    print(combined_df.head())
else:
    print("No valid data was retrieved.")
```
